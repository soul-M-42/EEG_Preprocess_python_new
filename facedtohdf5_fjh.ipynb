```python
import mne
import numpy as np
import h5py
import os
from pathlib import Path

# --- 1. 配置路径 ---
path_data = Path("/mnt/dataset4/fjh/FACED/Data/sub000/data.bdf")
path_evt = Path("/mnt/dataset4/fjh/FACED/Data/sub000/evt.bdf")
save_path = Path("sub000_test.h5")

# Labram (TUEV) 标准通道顺序 (23通道)
# 依据: make_TUEV.py 中的 chOrder_standard
LABRAM_CHANNELS = [
    'FP1', 'FP2', 'F3', 'F4', 'C3', 'C4', 'P3', 'P4', 'O1', 'O2', 
    'F7', 'F8', 'T3', 'T4', 'T5', 'T6', 'A1', 'A2', 'FZ', 'CZ', 
    'PZ', 'T1', 'T2'
]

# 视频 ID -> 情感标签 (0-8)
VIDEO_ID_TO_LABEL = {
    1:0, 2:0, 3:0,       # Anger
    4:1, 5:1, 6:1,       # Disgust
    7:2, 8:2, 9:2,       # Fear
    10:3, 11:3, 12:3,    # Sadness
    13:4, 14:4, 15:4, 16:4, # Neutral
    17:5, 18:5, 19:5,    # Amusement
    20:6, 21:6, 22:6,    # Inspiration
    23:7, 24:7, 25:7,    # Joy
    26:8, 27:8, 28:8     # Tenderness
}

TARGET_FREQ = 200

# --- 2. 加载与预处理 ---
# preload=True 必须开启以支持后续滤波
raw = mne.io.read_raw_bdf(path_data, preload=True, verbose=False)

# 滤波: 去工频 (50Hz) + 带通 (0.1-75Hz)
raw.notch_filter(freqs=50.0, verbose=False)
raw.filter(l_freq=0.1, h_freq=75.0, verbose=False)

# 加载标注
try:
    annot = mne.read_annotations(path_evt)
    raw.set_annotations(annot)
except Exception as e:
    raise RuntimeError(f"Failed to load annotations: {e}")

# 重采样至 200Hz
if raw.info['sfreq'] != TARGET_FREQ:
    raw.resample(TARGET_FREQ, verbose=False)

# --- 3. 解析事件与切片 ---
events, event_id_map = mne.events_from_annotations(raw, verbose=False)
id_to_desc = {v: k for k, v in event_id_map.items()}

extracted_data = []
state = 0   
curr_vid = None
t_start_idx = 0

for ev in events:
    idx = ev[0]
    desc = id_to_desc[ev[2]]

    # 状态机: 找ID -> 找101(Start) -> 找102(End)
    if state == 0: 
        if desc.isdigit() and 1 <= int(desc) <= 28:
            curr_vid = int(desc)
            state = 1
    
    elif state == 1:
        if desc == '101':
            t_start_idx = idx
            state = 2
        elif desc.isdigit() and 1 <= int(desc) <= 28:
             curr_vid = int(desc)

    elif state == 2:
        if desc == '102':
            t_end_idx = idx
            # 获取数据 (MNE自动转uV)
            chunk = raw.get_data(start=t_start_idx, stop=t_end_idx, units='uV')
            label_val = VIDEO_ID_TO_LABEL.get(curr_vid, -1)
            
            extracted_data.append({
                'data': chunk,
                'vid': curr_vid,
                'label': label_val
            })
            state = 0
            curr_vid = None

print(f"Extracted {len(extracted_data)} segments.")

# --- 4. 对齐通道并保存HDF5 ---
src_ch_upper = [ch.strip().upper() for ch in raw.ch_names]

with h5py.File(save_path, 'w') as f:
    for item in extracted_data:
        raw_chunk = item['data']
        vid = item['vid']
        label = item['label']

        n_samples = raw_chunk.shape[1]
        aligned_chunk = np.zeros((23, n_samples), dtype=np.float32)
        
        # 逐个通道对齐
        for i, target_ch in enumerate(LABRAM_CHANNELS):
            if target_ch in src_ch_upper:
                src_idx = src_ch_upper.index(target_ch)
                aligned_chunk[i, :] = raw_chunk[src_idx, :]
            # 找不到的通道 (如 T1, T2) 保持为 0
        
        # 创建 Group: sub000_vidXX
        grp_name = f"sub000_vid{vid}"
        grp = f.create_group(grp_name)
        dset = grp.create_dataset('eeg', data=aligned_chunk)

        # 写入关键属性
        dset.attrs['chOrder'] = LABRAM_CHANNELS
        dset.attrs['rsFreq'] = TARGET_FREQ
        dset.attrs['label'] = label
        dset.attrs['video_id'] = vid

print(f"Saved to {save_path}")
```

```python
import mne
import numpy as np
import h5py
import os
import json
from pathlib import Path
from tqdm import tqdm  # 用于显示进度条: pip install tqdm

# ================= 1. 全局配置 =================

# 原始数据根目录
DATA_ROOT = Path("/mnt/dataset4/fjh/FACED/Data")
OUTPUT_DIR = Path("/mnt/dataset4/fjh/FACED/FACED_labram")

# Labram (TUEV) 标准通道
LABRAM_CHANNELS = [
    'FP1', 'FP2', 'F3', 'F4', 'C3', 'C4', 'P3', 'P4', 'O1', 'O2', 
    'F7', 'F8', 'T3', 'T4', 'T5', 'T6', 'A1', 'A2', 'FZ', 'CZ', 
    'PZ', 'T1', 'T2'
]

# 视频ID -> 情感标签
VIDEO_ID_TO_LABEL = {
    1:0, 2:0, 3:0,       # Anger
    4:1, 5:1, 6:1,       # Disgust
    7:2, 8:2, 9:2,       # Fear
    10:3, 11:3, 12:3,    # Sadness
    13:4, 14:4, 15:4, 16:4, # Neutral
    17:5, 18:5, 19:5,    # Amusement
    20:6, 21:6, 22:6,    # Inspiration
    23:7, 24:7, 25:7,    # Joy
    26:8, 27:8, 28:8     # Tenderness
}

TARGET_FREQ = 200

# ================= 2. 辅助函数 =================

def get_cohort_mapping(sub_folder):
    """
    判断被试是否属于 Cohort 2 (sub061+)，需要特殊映射
    """
    # 简单的 ID 判断法
    try:
        sub_id = int(sub_folder.name.replace('sub', ''))
        if sub_id >= 61:
            return {
                'T7': 'T3', 'T8': 'T4',
                'P7': 'T5', 'P8': 'T6',
                'HEOL': 'A1', 'HEOR': 'A2'
            }
    except:
        pass
    return {}

def process_single_subject(sub_folder, output_dir):
    """
    读取单个被试文件夹 -> 预处理 -> 切片 -> 保存为单独的 H5
    """
    sub_name = sub_folder.name
    path_data = sub_folder / "data.bdf"
    path_evt = sub_folder / "evt.bdf"
    save_path = output_dir / f"{sub_name}.h5"

    # 1. 检查文件
    if not path_data.exists() or not path_evt.exists():
        return False, "Files missing"

    # 2. 加载 Raw
    try:
        raw = mne.io.read_raw_bdf(path_data, preload=True, verbose=False)
    except Exception as e:
        return False, f"Load Error: {e}"

    # 3. 预处理 (滤波)
    try:
        raw.notch_filter(freqs=50.0, verbose=False)
        raw.filter(l_freq=0.1, h_freq=75.0, verbose=False)
    except Exception as e:
        return False, f"Filter Error: {e}"

    # 4. 加载 Annotations
    try:
        annot = mne.read_annotations(path_evt)
        raw.set_annotations(annot)
    except Exception as e:
        return False, f"Annot Error: {e}"

    # 5. 重采样
    if raw.info['sfreq'] != TARGET_FREQ:
        raw.resample(TARGET_FREQ, verbose=False)

    # 6. 解析事件 & 切片
    try:
        events, event_id_map = mne.events_from_annotations(raw, verbose=False)
    except ValueError:
        return False, "No events found"

    id_to_desc = {v: k for k, v in event_id_map.items()}
    extracted_data = []
    
    state = 0; curr_vid = None; t_start_idx = 0

    for ev in events:
        idx = ev[0]; desc = id_to_desc[ev[2]]
        
        # 状态机逻辑
        if state == 0: 
            if desc.isdigit() and 1 <= int(desc) <= 28:
                curr_vid = int(desc); state = 1
        elif state == 1:
            if desc == '101':
                t_start_idx = idx; state = 2
            elif desc.isdigit() and 1 <= int(desc) <= 28:
                 curr_vid = int(desc)
        elif state == 2:
            if desc == '102':
                t_end_idx = idx
                # 获取数据 (uV)
                chunk = raw.get_data(start=t_start_idx, stop=t_end_idx, units='uV')
                label = VIDEO_ID_TO_LABEL.get(curr_vid, -1)
                extracted_data.append({'data': chunk, 'vid': curr_vid, 'label': label})
                state = 0; curr_vid = None

    if len(extracted_data) == 0:
        return False, "No segments extracted"

    # 7. 对齐通道 & 保存 H5
    mapping = get_cohort_mapping(sub_folder)
    src_ch_upper = [ch.strip().upper() for ch in raw.ch_names]

    with h5py.File(save_path, 'w') as f:
        for item in extracted_data:
            raw_chunk = item['data'].astype(np.float32) # 压缩为 float32
            vid = item['vid']
            label = item['label']

            n_samples = raw_chunk.shape[1]
            aligned_chunk = np.zeros((23, n_samples), dtype=np.float32)
            
            for i, target_ch in enumerate(LABRAM_CHANNELS):
                src_idx = -1
                # A. 直接匹配
                if target_ch in src_ch_upper:
                    src_idx = src_ch_upper.index(target_ch)
                # B. 映射匹配 (针对 sub061+)
                elif target_ch in mapping.values():
                    origs = [k for k, v in mapping.items() if v == target_ch]
                    for orig in origs:
                        if orig.upper() in src_ch_upper:
                            src_idx = src_ch_upper.index(orig.upper())
                            break
                
                if src_idx != -1:
                    aligned_chunk[i, :] = raw_chunk[src_idx, :]
            
            # 写入 Group
            grp_name = f"vid{vid}" # 文件名已经是subXXX了，这里只需 vid
            grp = f.create_group(grp_name)
            dset = grp.create_dataset('eeg', data=aligned_chunk)
            
            dset.attrs['chOrder'] = LABRAM_CHANNELS
            dset.attrs['rsFreq'] = TARGET_FREQ
            dset.attrs['label'] = label
            dset.attrs['video_id'] = vid

    return True, f"Saved {len(extracted_data)} segs"

# ================= 3. 批量执行逻辑 =================

if __name__ == "__main__":
    
    # 1. 确保输出目录存在
    if not OUTPUT_DIR.exists():
        OUTPUT_DIR.mkdir(parents=True)
        print(f"Created directory: {OUTPUT_DIR}")
    
    # 2. 获取所有被试目录
    sub_folders = sorted(list(DATA_ROOT.glob("sub*")))
    print(f"Found {len(sub_folders)} subjects in {DATA_ROOT}")
    
    # 3. 进度条循环
    pbar = tqdm(sub_folders)
    for sub_folder in pbar:
        if not sub_folder.is_dir(): continue
        
        # 显示当前正在处理谁
        pbar.set_description(f"Processing {sub_folder.name}")
        
        # 执行处理
        success, msg = process_single_subject(sub_folder, OUTPUT_DIR)
        
        if not success:
            print(f"\n⚠️ {sub_folder.name}: {msg}")

    print("\n✅ 所有处理完成！")
```

```python
import h5py
import os
import random
from pathlib import Path
from sklearn.model_selection import train_test_split
from tqdm import tqdm

# ================= 配置区域 =================
# 1. 文件夹路径 
DATA_DIR = Path("/mnt/dataset4/fjh/FACED/FACED_labram").resolve()

# 2. 输出 txt 文件的保存目录
OUTPUT_DIR = Path("./FACED_splits")

# 3. 划分比例 (Train : Val : Test)
TRAIN_RATIO = 0.8
VAL_RATIO = 0.1 

SEED = 42  # 固定随机种子，保证每次运行分得一样

# ================= 主程序 =================

def generate_splits():
    if not DATA_DIR.exists():
        print(f"❌ 错误：找不到数据目录 {DATA_DIR}")
        return

    # 创建输出目录
    if not OUTPUT_DIR.exists():
        OUTPUT_DIR.mkdir(parents=True)

    # 1. 获取所有被试文件 
    all_files = sorted(list(DATA_DIR.glob("sub*.h5")))
    subject_files = [f for f in all_files if f.is_file()]
    
    total_subs = len(subject_files)
    print(f"📊 扫描到 {total_subs} 个被试文件。")
    
    if total_subs == 0:
        print("⚠️ 没找到 .h5 文件，请检查路径！")
        return

    # 2. 按被试划分 (防止数据泄漏)
    # 先分出 Train + (Val+Test)
    train_subs, temp_subs = train_test_split(
        subject_files, 
        train_size=TRAIN_RATIO, 
        random_state=SEED,
        shuffle=True
    )
    
    remaining_ratio = 1.0 - TRAIN_RATIO
    relative_val_size = VAL_RATIO / remaining_ratio

    # 再把 (Val+Test) 对半分为 Val 和 Test
    val_subs, test_subs = train_test_split(
        temp_subs, 
        train_size=relative_val_size, 
        random_state=SEED,
        shuffle=True
    )

    print(f"   🔹 训练集: {len(train_subs)} 人")
    print(f"   🔹 验证集: {len(val_subs)} 人")
    print(f"   🔹 测试集: {len(test_subs)} 人")

    # 3. 写入 txt 索引文件
    # 格式: H5文件绝对路径,内部Group名,标签
    
    def write_index_file(subset_name, file_list):
        output_path = OUTPUT_DIR / f"{subset_name}.txt"
        lines = []
        
        print(f"⚡ 正在生成 {subset_name}.txt ...")
        
        for h5_path in tqdm(file_list):
            try:
                # 只读模式打开 H5，快速扫描里面的 Group
                with h5py.File(h5_path, 'r') as f:
                    # f.keys() 得到 ['vid1', 'vid2', ..., 'vid28']
                    for group_name in f.keys():
                        # 读取 label 属性
                        # 注意：我们之前存的时候是 dset.attrs['label']
                        # dset 是 f[group_name]['eeg']
                        label = f[group_name]['eeg'].attrs['label']
                        
                        # 构造一行记录
                        # 格式: /path/to/sub001.h5,vid1,0
                        line = f"{h5_path},{group_name},{label}"
                        lines.append(line)
            except Exception as e:
                print(f"   ⚠️ 读取 {h5_path.name} 失败: {e}")

        # 写入文件
        with open(output_path, 'w') as f_out:
            f_out.write('\n'.join(lines))
        
        print(f"✅ {subset_name}.txt 生成完毕！共包含 {len(lines)} 个样本。")

    # 执行写入
    write_index_file("train", train_subs)
    write_index_file("val", val_subs)
    write_index_file("test", test_subs)

if __name__ == "__main__":
    generate_splits()
```
